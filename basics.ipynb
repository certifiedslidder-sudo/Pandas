### Concatenate and merge dataframe (JOINS)

```python
df1 = pd.DataFrame({'ID': [1,2,3],'Name' : ['A','B','C']})
df1
```

```python
df2 = pd.DataFrame({'ID': [1,2,2,4],'Score' : [88,96,77,79]})
df2
```

```python
pd.concat([df1,df2], axis =0)   # top on top / row level / vertical
# nulls the value that don't match
```

```python
pd.concat([df1,df2], axis =1)   # side on side
```

# MERGE

```python
# INNER (default)= Keeps only matching rows based on the join keys.
#Rows without a match in both DataFrames are dropped.
```

```python
## OUTER = Keeps all rows from both DataFrames.
#Missing values are filled with NaN.
```

```python
# LEFT = Keeps all rows from the left DataFrame.
#Adds matching rows from the right DataFrame; unmatched rows get NaN.
```

```python
# RIGHT = Keeps all rows from the right DataFrame.
#Adds matching rows from the left DataFrame; unmatched rows get NaN.
```

```python
#merge = join
pd.merge(df1, df2 , how = 'inner', on = 'ID')

# SYNTAX=    (left: DataFrame | Series[Any], right: DataFrame | Series[Any], how: MergeHow = "inner", on: Label)
```

```python
#merge = join
pd.merge(df1, df2 , how = 'outer', on = 'ID')
```

```python
#merge = join
pd.merge(df1, df2 , how = 'left', on = 'ID')

# SYNTAX=    (left: DataFrame | Series[Any], right: DataFrame | Series[Any], how: MergeHow = "inner", on: Label)
```

```python
#merge = join
pd.merge(df1, df2 , how = 'right', on = 'ID')
```

```python
pd.merge(df1,df2,how ='inner', left_on ='ID', right_on='ID')
```
